## 1. O que é o Pandas?

Pandas é uma biblioteca Python que fornece estruturas de dados de alta performance e ferramentas de análise de dados.

### Principais estruturas:

1. **Series**: Array unidimensional (uma coluna)
2. **DataFrame**: Estrutura bidimensional (tabela com linhas e colunas)

### Por que usar Pandas?
- ✅ Fácil manipulação de dados tabulares
- ✅ Leitura/escrita de diversos formatos (CSV, Excel, SQL, JSON)
- ✅ Limpeza e preparação de dados
- ✅ Análise estatística rápida
- ✅ Integração com outras bibliotecas (NumPy, Matplotlib)

## 3. Series - A Estrutura Unidimensional

Uma Series é como uma coluna de dados. Pode ser criada a partir de uma lista, dicionário ou array NumPy.

````python
# Criando uma Series a partir de uma lista
notas = pd.Series([8, 9, 7, 10, 6])
print(notas)
print("\n" + "="*40)
print(f"Média: {notas.mean()}")
print(f"Máximo: {notas.max()}")
print(f"Mínimo: {notas.min()}")
````

## 4. DataFrame - A Estrutura Bidimensional

Um DataFrame é como uma planilha ou tabela SQL. É a estrutura mais usada no Pandas.

````python

import matplotlib.pyplot as plt







dados = {
    'Nome': ['Ana', 'João', 'Maria', 'Pedro', 'Carla'],
    'Idade': [25, 30, 28, 35, 22],
    'Cidade': ['São Paulo', 'Rio', 'Belo Horizonte', 'Curitiba', 'Salvador'],
    'Nota': [8.5, 9.0, 7.5, 9.5, 8.0]
}

df = pd.DataFrame(dados)
plt.bar(df.Nome,df.Nota)
plt.show()
df
````

## 5. Visualizando Dados

O Pandas oferece várias formas de visualizar os dados:

````python
# Visualizando as primeiras linhas
print("Primeiras 3 linhas:")
print(df.head(3))

print("\nÚltimas 2 linhas:")
print(df.tail(2))

print("\nInformações do DataFrame:")
print(df.info())

print("\nEstatísticas descritivas:")
print(df.describe())
````

## 6. Selecionando Colunas

Você pode selecionar colunas de várias formas:

````python
# Selecionando uma coluna (retorna uma Series)
print("Coluna Nome:")
print(df['Nome'])

print("\nMúltiplas colunas (retorna um DataFrame):")
print(df[['Nome', 'Nota']])
````

## 7. Adicionando Nova Coluna

É muito fácil adicionar novas colunas a um DataFrame:

````python
# Adicionando uma coluna de status (aprovado/reprovado)
df['Status'] = df['Nota'].apply(lambda x: 'Aprovado' if x >= 7 else 'Reprovado')
print(df)
````

### Exercício 3: Series e Métodos Básicos
Crie uma Series com as idades de 10 pessoas. Calcule e imprima:
- Média das idades
- Mediana
- Desvio padrão
- Idade máxima e mínima
- Número total de elementos

````python
idades = pd.Series([54, 23, 98, 18, 42, 64, 97, 27, 35])

print(f"Média: {idades.mean():.2f}")
print(f"Mediana: {idades.median():.2f}")
print(f"Desvio padrão: {idades.std():.2f}")
print(f"Número de elementos: {idades.count()}")
````
